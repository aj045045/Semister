```python
import numpy as np
import matplotlib.pyplot as plt
```

## Sigmoid activation function

```python
def sigmoid(x):
    return 1 / ( 1 + np.exp(-x))

def sigmoid_derivative(a):
    return a * (1 - a )
```

```python
x = np.linspace(-10,10,1000)
s1 = sigmoid(x)
s2 = sigmoid_derivative(s1)

plt.plot(x,s1,label='sigmoid')
plt.plot(x,s2,label='derivative')
plt.legend()
plt.show()
```

## Tanh activation function

```python
def tanh(x):
    return (np.exp(x) - np.exp(-x) ) / (np.exp(x) + np.exp(-x))

def tanh_derivate(x):
    return 1-x**2
```

```python
x = np.linspace(-10,10,1000)
s1 = tanh(x)
s2 = tanh_derivate(s1)
plt.plot(x,s1,label='tanh')
plt.plot(x,s2,label='derivation')
plt.legend()
plt.show()
```

## ReLU Activation function

```python
def ReLU(x):
    data = [max(0,value) for value in x]
    return np.array(data,dtype=float)

def ReLU_derivative(x):
    data = [1 if value > 0 else 0 for value in x]
    return np.array(data,dtype=float)
```

```python
x = np.linspace(-10,10,1000)
s1 = ReLU(x)
s2 = ReLU_derivative(s1)

plt.plot(x,s1,label='ReLU')
plt.plot(x,s2,label='derivation')
plt.legend()
plt.grid(True)
plt.show()
```

## Leaky ReLU

```python
def Leaky_ReLU(x):
    data = [max(0.05*value,value) for value in x]
    return np.array(data,dtype=float)

def Leaky_ReLU_derivative(x):
    data = [1 if value > 0 else 0.05 for value in x]
    return np.array(data,dtype=float)
```

```python
x = np.linspace(-10,10,1000)
s1 = Leaky_ReLU(x)
s2 = Leaky_ReLU_derivative(s1)
plt.plot(x,s1,label='Leaky ReLU')
plt.plot(x,s2,label='derivation')
plt.legend()
plt.grid()
plt.show()
```
