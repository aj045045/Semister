````python
import numpy as np
import matplotlib.pyplot as plt
````

## 1.WAP of Activation function from scratch. (Sigmoid, Tanh, ReLU, Leaky ReLU)

````python
# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
    
def sigmoid_derivate(x):
    return x * (1 - x)

x = np.linspace(-10, 10, 1000)
s1 = sigmoid(x)
s2 = sigmoid_derivate(s1)

plt.plot(x,s1,label='sigmoid')
plt.plot(x,s2,label='sigmoid derivate')
plt.legend()
plt.show()
````

````python
# Tanh
def tanh(x):
    return (np.exp(x) - np.exp(-x) ) / ( np.exp(x) + np.exp(-x))

def tanh_derivate(x):
    return 1 - x**2

x = np.linspace(-10, 10, 1000)
s1 = tanh(x)
s2 = tanh_derivate(s1)

plt.plot(x,s1,label='tanh')
plt.plot(x,s2,label='tanh derivate')
plt.legend()
plt.show()
````

````python

# ReLU
def relu(x):
    data = [max(0,value) for value in x]
    return np.array(data,dtype=float)

def relu_derivate(x):
    data = [1 if value > 0 else 0 for value in x]
    return np.array(data,dtype=float)

x = np.linspace(-10, 10, 1000)
s1 = relu(x)
s2 = relu_derivate(s1)

plt.plot(x,s1,label='relu')
plt.plot(x,s2,label='relu derivate')
plt.legend()
plt.grid()
plt.show()
````

````python
# Leaky ReLU
def leaky_relu(x):
    data = [max(0.05*value,value) for value in x]
    return np.array(data,dtype=float)

def leaky_relu_derivate(x):
    data = [1 if value > 0 else 0.05 for value in x]
    return np.array(data,dtype=float)
    
x = np.linspace(-10, 10, 1000)
s1 = leaky_relu(x)
s2 = leaky_relu_derivate(s1)

plt.plot(x,s1,label='relu')
plt.plot(x,s2,label='relu derivate')
plt.legend()
plt.grid()
plt.show()
````

## 2.WAP of ANN model on MNIST digit classification and Fashion MNIST from keras dataset.

````python
# Import necessary libraries
import tensorflow as tf
from tensorflow import keras
from keras import layers
import matplotlib.pyplot as plt

# Load MNIST and Fashion-MNIST datasets
mnist = keras.datasets.mnist
fashion_mnist = keras.datasets.fashion_mnist

# Load and split data into training and test sets
(x_train_mnist, y_train_mnist), (x_test_mnist, y_test_mnist) = mnist.load_data()
(x_train_fashion, y_train_fashion), (x_test_fashion, y_test_fashion) = fashion_mnist.load_data()

# Normalize pixel values to range [0, 1] (for faster convergence)
x_train_mnist, x_test_mnist = x_train_mnist / 255.0, x_test_mnist / 255.0
x_train_fashion, x_test_fashion = x_train_fashion / 255.0, x_test_fashion / 255.0

# Flatten images from 28x28 to 784 (for ANN input)
x_train_mnist = x_train_mnist.reshape(-1, 784)
x_test_mnist = x_test_mnist.reshape(-1, 784)
x_train_fashion = x_train_fashion.reshape(-1, 784)
x_test_fashion = x_test_fashion.reshape(-1, 784)

# Define an ANN model
def create_ann_model():
    model = keras.Sequential([
        layers.Input(shape=(784,), name="pixels"),
        layers.Dense(512, activation='relu'),  # First hidden layer with 512 neurons
        layers.Dense(256, activation='relu'),  # Second hidden layer with 256 neurons
        layers.Dense(128, activation='relu'),  # Third hidden layer with 128 neurons
        layers.Dense(10, activation='softmax')  # Output layer (10 classes)
    ])
    return model

# 9️⃣ Plot training history
def plot_history(hist):
    plt.figure(figsize=(10, 4))
    plt.plot(hist.history['loss'], label='Train Loss')
    plt.plot(hist.history['val_loss'], label='Val Loss')
    plt.title('Loss over Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()


# Compile and Train the Model
def compile_and_train(model, x_train, y_train, x_test, y_test, dataset_name):
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    
    print(f"\nTraining on {dataset_name} dataset...")
    history = model.fit(x_train, y_train, epochs=10, batch_size=64, validation_data=(x_test, y_test))
    plot_history(history)
    # Evaluate on test set
    test_loss, test_acc = model.evaluate(x_test, y_test)
    print(f"Test accuracy on {dataset_name}: {test_acc:.4f}")

# Train and evaluate the model on MNIST
model_mnist = create_ann_model()
compile_and_train(model_mnist, x_train_mnist, y_train_mnist, x_test_mnist, y_test_mnist, "MNIST")

# Train and evaluate the model on Fashion-MNIST
model_fashion = create_ann_model()
compile_and_train(model_fashion, x_train_fashion, y_train_fashion, x_test_fashion, y_test_fashion, "Fashion-MNIST")
````
