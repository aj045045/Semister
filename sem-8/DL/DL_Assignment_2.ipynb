# 1. WAP of CNN from scratch on image classification.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Load and preprocess data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train = x_train.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
x_test = x_test.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

# Build CNN from scratch using TensorFlow low-level API
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(8, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(16, kernel_size=(3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, batch_size=64, epochs=3, validation_split=0.1)

test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy:", test_acc)
```

# 2. WAP of RNN from scratch on stock data.

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# Load stock data
df = pd.read_csv('./dataset/GOOGL.csv')
data = df['Close'].values.reshape(-1, 1)
# Normalize
scaler = MinMaxScaler()
data = scaler.fit_transform(data)

# Prepare sequences
def create_dataset(data, time_step=60):
    x, y = [], []
    for i in range(len(data) - time_step):
        x.append(data[i:i+time_step])
        y.append(data[i+time_step])
    return np.array(x), np.array(y)

time_step = 60
X, y = create_dataset(data, time_step)
X = X.reshape(X.shape[0], X.shape[1], 1)  # RNN needs 3D input

# Split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# RNN model from scratch
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(time_step,1)),
    tf.keras.layers.SimpleRNN(50, activation='tanh', return_sequences=False),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(1)
])

# 9️⃣ Plot training history
def plot_history(hist):
    plt.figure(figsize=(10, 4))
    plt.plot(hist.history['loss'], label='Train Loss')
    plt.plot(hist.history['val_loss'], label='Val Loss')
    plt.title('Loss over Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()


model.compile(optimizer='adam', loss='mse')
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.1)
plot_history(history)

# Predict and plot
predicted = model.predict(X_test)
predicted = scaler.inverse_transform(predicted)
actual = scaler.inverse_transform(y_test.reshape(-1, 1))

plt.figure(figsize=(10, 5))
plt.plot(actual, label="Actual Price")
plt.plot(predicted, label="Predicted Price")
plt.legend()
plt.title("GOOGL Stock Price Prediction using RNN")
plt.xlabel("Time")
plt.ylabel("Price")
plt.show()
```

# 3. WAP of LSTM from scratch on Time series data.

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
from tensorflow.keras.callbacks import EarlyStopping
# Load data
df = pd.read_csv('./dataset/Electric_Production.csv')
data = df['IPG2211A2N'].values
if data.ndim == 1:
    data = data.reshape(-1, 1)

# Normalize
scaler = MinMaxScaler()
data = scaler.fit_transform(data)

# Sequence creation
def create_dataset(data, time_step=12):
    x, y = [], []
    for i in range(len(data) - time_step):
        x.append(data[i:i+time_step])
        y.append(data[i+time_step])
    return np.array(x), np.array(y)

time_step = 12
X, y = create_dataset(data, time_step)
X = X.reshape(X.shape[0], X.shape[1], 1)  # LSTM input shape

# Train/test split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# Build LSTM model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(time_step, 1)),
    tf.keras.layers.LSTM(16, return_sequences=False),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1)
])
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
model.compile(optimizer='adam', loss='mse')
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.1,callbacks=[early_stop] )
plot_history(history)

# Prediction
predicted = model.predict(X_test)
predicted = scaler.inverse_transform(predicted)
actual = scaler.inverse_transform(y_test.reshape(-1, 1))

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(actual, label="Actual")
plt.plot(predicted, label="Predicted")
plt.title("Electric Production Forecast using LSTM")
plt.xlabel("Time")
plt.ylabel("Production")
plt.legend()
plt.show()
```

# 4. WAP of GRU from scratch on Time series data as well as on generating text.

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# Load beer production data
df = pd.read_csv('./dataset/monthly-beer-production-in-austr.csv')
data = df['Monthly beer production'].values.reshape(-1, 1)

# Normalize
scaler = MinMaxScaler()
data = scaler.fit_transform(data)

# Create sequences
def create_dataset(data, time_step=12):
    x, y = [], []
    for i in range(len(data) - time_step):
        x.append(data[i:i+time_step])
        y.append(data[i+time_step])
    return np.array(x), np.array(y)

time_step = 12
X, y = create_dataset(data, time_step)
X = X.reshape(X.shape[0], X.shape[1], 1)

# Split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

# GRU model
model = tf.keras.Sequential([
    tf.keras.layers.GRU(64, activation='tanh', input_shape=(time_step, 1)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=20, batch_size=32, validation_split=0.1)

# Predict
predicted = model.predict(X_test)
predicted = scaler.inverse_transform(predicted)
actual = scaler.inverse_transform(y_test.reshape(-1, 1))

# Plot
plt.figure(figsize=(10, 5))
plt.plot(actual, label='Actual')
plt.plot(predicted, label='Predicted')
plt.title("Beer Production Forecast using GRU")
plt.xlabel("Time")
plt.ylabel("Production")
plt.legend()
plt.show()
```

```python
import tensorflow as tf
import numpy as np

# Example text
text = "deep learning is changing the future of ai. "
chars = sorted(set(text))
char2idx = {c: i for i, c in enumerate(chars)}
idx2char = np.array(chars)

# Prepare training data
seq_length = 10
step = 1
X = []
y = []
for i in range(0, len(text) - seq_length, step):
    X.append([char2idx[c] for c in text[i:i+seq_length]])
    y.append(char2idx[text[i+seq_length]])

X = tf.keras.utils.to_categorical(X, num_classes=len(chars))
y = tf.keras.utils.to_categorical(y, num_classes=len(chars))

# GRU model
model = tf.keras.Sequential([
    tf.keras.layers.GRU(128, input_shape=(seq_length, len(chars))),
    tf.keras.layers.Dense(len(chars), activation='softmax')
])
model.compile(loss='categorical_crossentropy', optimizer='adam')
model.fit(X, y, batch_size=1, epochs=50, verbose=0)

# Generate text
def generate_text(model, start_str, length=100):
    result = start_str
    input_seq = [char2idx[c] for c in start_str]
    
    for _ in range(length):
        x_pred = tf.keras.utils.to_categorical([input_seq], num_classes=len(chars))
        pred = model.predict(x_pred, verbose=0)[0]
        next_idx = np.argmax(pred)
        next_char = idx2char[next_idx]
        result += next_char
        input_seq = input_seq[1:] + [next_idx]
    return result

print("Generated text:\n", generate_text(model, "deep learn"))
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Input,Dropout
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import datetime
import visualkeras
import seaborn as sns
from PIL import ImageFont

# 1️⃣ Set random seed for reproducibility
tf.random.set_seed(42)

# 2️⃣ Load and preprocess MNIST data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train = x_train.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
x_test = x_test.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
y_train_cat = to_categorical(y_train, num_classes=10)
y_test_cat = to_categorical(y_test, num_classes=10)

print(f"Training samples: {x_train.shape}, Testing samples: {x_test.shape}")

# 3️⃣ Define CNN Model Architecture
model = Sequential([
    Input(shape=(28,28,1)),
    Conv2D(8, kernel_size=(3, 3), activation='relu',name="Conv1"),
    MaxPooling2D((2, 2), name="MaxPool1"),
    Dropout(0.2),
    Conv2D(16, kernel_size=(3, 3), activation='relu', name="Conv2"),
    MaxPooling2D((2, 2), name="MaxPool2"),
    Dropout(0.2),
    Flatten(name="Flatten"),
    Dense(128, activation='relu', name="Dense1"),
    Dense(10, activation='softmax', name="Output")
])

font = ImageFont.load_default()
visual_img = visualkeras.layered_view(model,font=font, legend=True)

plt.figure(figsize=(12, 6))
plt.imshow(visual_img)
plt.axis('off')
plt.title('Model Architecture')
plt.show()


# 4️⃣ Print model summary
print("\n📦 Model Summary:")
model.summary()

# 5️⃣ Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

early_stopping = EarlyStopping(
    monitor='val_loss',   # metric to monitor
    patience=3,           # number of epochs with no improvement to wait
    restore_best_weights=True  # restore weights from the epoch with the best value
)

# 7️⃣ Train the model
history = model.fit(
    x_train, y_train_cat,
    batch_size=64,
    epochs=10,
    validation_split=0.1,
    callbacks=[early_stopping],
    verbose=2
)

# 8️⃣ Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test_cat, verbose=0)
print(f"\n✅ Test Accuracy: {test_acc:.4f} | Test Loss: {test_loss:.4f}")

# 9️⃣ Plot training history
def plot_history(hist):
    plt.figure(figsize=(10, 4))

    # Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(hist.history['accuracy'], label='Train Acc')
    plt.plot(hist.history['val_accuracy'], label='Val Acc')
    plt.title('Accuracy over Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()

    # Loss
    plt.subplot(1, 2, 2)
    plt.plot(hist.history['loss'], label='Train Loss')
    plt.plot(hist.history['val_loss'], label='Val Loss')
    plt.title('Loss over Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()

plot_history(history)

# 🔟 Predict on a sample image
sample_index = 5
sample_img = x_test[sample_index]
prediction = model.predict(np.expand_dims(sample_img, axis=0))
predicted_label = np.argmax(prediction)
actual_label = y_test[sample_index]

plt.imshow(sample_img.squeeze(), cmap='gray')
plt.title(f"Predicted: {predicted_label} | Actual: {actual_label}")
plt.axis('off')
plt.show()
    
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import zscore

# 1. Load the data
df = pd.read_csv('./dataset/GOOGL.csv')

# 2. Optional: See the columns
print(df.columns)

# 3. Drop non-numeric columns (Z-score only works on numeric features)
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# 4. Plot original data
plt.figure(figsize=(12, 6))
sns.boxplot(data=numeric_df)
plt.title('Before Removing Outliers')
plt.xticks(rotation=45)
plt.show()

# 5. Compute Z-scores
z_scores = zscore(numeric_df)

# 6. Filter: Keep only rows where all z-scores are < 3
filtered_entries = (abs(z_scores) < 2).all(axis=1)
df_clean = df[filtered_entries]

# 7. Plot cleaned data
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean.select_dtypes(include=['float64', 'int64']))
plt.title('After Removing Outliers (Z-Score Method)')
plt.xticks(rotation=45)
plt.show()
```
