```python
import os
import tensorflow as tf
import mlflow
import mlflow.tensorflow

# ========== Safe MLflow Autolog ==========
mlflow.tensorflow.autolog()
mlflow.set_experiment("ML_Flow_DL_Test")
mlflow.set_tracking_uri('http://localhost:5000')
# Optional: Force CPU usage (recommended in Jupyter if GPU issues arise)
# os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

# ========== Configuration ==========
EPOCHS = 20
MODEL_DIR = "model_output"
MODEL_PATH = os.path.join(MODEL_DIR, "mnist_model.keras")


# ========== Load and Prepare Data ==========
def load_data():
    (x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
    x_train = x_train.astype("float32") / 255.0
    x_test = x_test.astype("float32") / 255.0
    return (x_train, y_train), (x_test, y_test)

# ========== Model Definition ==========
def build_model():
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(28, 28)),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(10, activation='softmax')
    ])
    return model

# ========== Training Logic ==========
def train_and_evaluate():
    (x_train, y_train), (x_test, y_test) = load_data()

    with mlflow.start_run(run_name="MNIST_Training_TF"):
        model = build_model()
        model.compile(optimizer='adam',
                      loss='sparse_categorical_crossentropy',
                      metrics=['accuracy'])

        # ✅ Eager execution is ON by default — keep it that way
        model.fit(x_train, y_train, epochs=EPOCHS, validation_split=0.1)

        test_loss, test_accuracy = model.evaluate(x_test, y_test)

        mlflow.log_metric("test_loss", test_loss)
        mlflow.log_metric("test_accuracy", test_accuracy)

        os.makedirs(MODEL_DIR, exist_ok=True)
        model.save(MODEL_PATH)
        mlflow.log_artifact(MODEL_PATH)

        mlflow.set_tags({
            "framework": "TensorFlow",
            "project": "MNIST Classification",
            "run_type": "autolog + eager",
        })

# ========== Run ==========
if __name__ == "__main__":
    train_and_evaluate()
```
